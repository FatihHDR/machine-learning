# MAIN TOPICS

Why should we study Artificial Neural Networks? Artificial Neural Networks are the foundation of many artificial intelligence technologies we use every day, such as face recognition, recommendation systems, automatic language translators, and others. Artificial Neural Networks have an extraordinary ability to recognize patterns from complex data.

## A. ARTIFICIAL NEURAL NETWORK (ANN)

Artificial Neural Network or neural network is a machine learning model that can help computers make decisions with a working method similar to the human brain. This process mimics how biological neurons work together to identify phenomena, consider options, up to the stage of decision-making or conclusion.

## B. BASIC STRUCTURE OF ARTIFICIAL NEURAL NETWORK (ANN)

Neural-Networks-Architecture.png

Artificial Neural Networks consist of several layers, with three main types:

**1. Input layer:** This layer is responsible for receiving raw data or features in the dataset. If we want to create face recognition, the input can be a face image.

**2. Hidden layer:** This layer processes data from the input layer. The more hidden layers there are, the more complex the network's patterns will be to learn and understand. Each hidden layer will transform into more complex and abstract information.

**3. Output layer:** This layer will produce the final decision or prediction. The output layer translates the information processed by the hidden layer into an understandable and usable form. For example, after processing an image, the output layer will decide whether the result is a cat image or a dog image.

## C. TRAINING PROCESS

1_ZXAOUqmlyECgfVa81Sr6Ew.webp

The training (learning) process of an artificial neural network is an iterative process where calculations are performed forwards and backwards through each layer in the network until the loss function is minimized. The entire series of processes is divided into three parts:

**1. Forward Propagation**

Artificial Neural Networks are made up of several neurons stacked into several layers. These layers are connected by network parameters (represented by arrows). These parameters are weights and biases. Weights will control the importance of each input. While bias will determine how easily neurons are activated.

The first thing to do is to initialize the network parameters with random non-zero values for weights and biases. Based on the input values and the initialized values, the linear function of the neuron and the activation function of the neuron will be calculated. This calculation will occur throughout the network and will output the final result of forward propagation in the first iteration.

In forward propagation, calculations are performed from the input layer to the output layer (left to right) through the network.

**2. Loss Function Calculation**

The output of forward propagation is called the predicted value. This value must be compared with the real value to measure the performance of the artificial neural network. This is where the loss function (also called the objective function or cost function) plays a role.

The loss function calculates the loss score between the predicted value and the actual value. This is also known as the error of a model. The loss function measures how well the model performs at each iteration. The loss function is used as feedback for parameter improvements in the backward propagation process. The ideal value of the loss function is 0. The closer to 0, the better the model's prediction will be.

**3. Backward Propagation**

In the first iteration, the predicted value will be far from the truth value. This is due to the arbitrary setting of parameters (weights and biases). These values are not optimal. Therefore, it is necessary to update the parameter values to minimize the loss function. The parameter update process is carried out using an optimizer that applies backward propagation. The goal of the optimizer is to find the global minimum point of the loss function.

In the backward propagation process, the gradient of the loss function with respect to the model parameters is calculated using the chain rule. This gradient indicates the direction to update the model parameters. Keras (a library in Python) automatically calculates these derivatives through automatic differentiation. Calculations are performed from the output layer to the input layer (from right to left).

## D. BATCH SIZE AND EPOCH

batch-vs-epoch.png

When dealing with the training and testing process, there are generally parameters that we can tune (adjust), including epoch and batch size.

Batch size is the number of data samples that typically pass through a neural network at one time. Batch size determines the number of samples to work on before updating the internal model parameters. For example, we have a dataset containing 1 million image samples for training. Although the neural network we use is not too complex, 1 million training data is still too much and heavy to process simultaneously.

Epoch is a hyperparameter that determines how many times the algorithm works through the entire dataset both forwards and backwards. One epoch is reached when all batches have successfully passed through the neural network once.

## E. TYPES OF ARTIFICIAL NEURAL NETWORKS

There are various types of Artificial Neural Network (ANN) architectures, including:

**1. Feedforward Neural Network (FNN):** The simplest type of artificial neural network where information flows in one direction from the input layer, through the hidden layer, and to the output layer. There are no loops or feedback in this network.

**2. Convolutional Neural Network (CNN):** Specifically designed for processing grid data such as images. CNNs use convolutional layers that can recognize complex visual patterns by applying filters to the input.

**3. Recurrent Neural Network (RNN):**
Has recurrent connections that allow information to persist along a sequence of data. This makes RNNs very useful for data that has temporal or sequential dependencies.

**4. Long Short-Term Memory (LSTM):** A special type of RNN designed to overcome the vanishing gradient problem by using memory cells that can retain information for longer periods.

# SIMPLE SIMULATION

Let's create a simple simulation of an Artificial Neural Network (ANN) for a binary classification task using Python and TensorFlow/Keras. Here are the steps:

**1. Data Preparation**

We will use a simple dataset like `make_moons` from sklearn to create easily understandable data. First, we import the necessary libraries.

````python
# Import library yang digunakan
# import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
````

Next, we use the `make_moons` function from the `sklearn.datasets` library to generate a synthetic dataset commonly used for classification problems.

````python
# Generate dataset
X, y = make_moons(n_samples=1000, noise=0.2, random_state=42)
````

Then, we split the data into training and test data.

````python
# Split dataset menjadi training dan testing set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
````

Next, perform sufficient preprocessing. In this simulation, we apply normalization to the data using "StandardScaler" to ensure that the features in the dataset have a similar scale, which can certainly improve the performance of machine learning algorithms.

````python
# Standarisasi data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
````

Finally, before building the Model, let's look at the data distribution using a Scatter Plot.

````python
# Plotting data
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap=plt.cm.Paired)
plt.title('Scatter Plot of Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
````

**2. Building the Model**

The model used in this simulation is a simple Feedforward Neural Network (FNN) built using the Keras framework from TensorFlow. First, let's import the libraries we need.

````python
# Import library yang akan digunakan
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
````

Next, let's build the model.

````python
# Membangun model
model = Sequential([
    Dense(10, input_shape=(2,), activation='relu'),
    Dense(10, activation='relu'),
    Dense(1, activation='sigmoid')
])
````

This model consists of three layers: an input layer with 10 neurons that receives two input features and uses the ReLU activation function, a hidden layer with 10 neurons that also uses the ReLU activation function, and an output layer with 1 neuron that uses the sigmoid activation function for binary classification. This model is designed to classify data into two classes based on its input features.
After the model has been created, let's compile the model by defining the optimizer, loss, and metrics parameters.

````python
# Compile model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
````

Here is an explanation of each of the parameters used:
1. Optimizer: The optimizer function is an algorithm used to adjust and update your model's parameters during training. Its goal is to minimize the value of the loss function by adjusting the weights and biases in the model. Some popular optimizers in machine learning include SGD (Stochastic Gradient Descent), Adam, RMSprop, and many more. Each optimizer has its own advantages and disadvantages and may be suitable for certain types of problems and data.
2. Loss: The loss function is a measure of how well or poorly your model predicts the training data compared to the actual labels. Its goal is to measure how large the prediction error of the model is. The choice of loss function should be tailored to the type of problem you are facing. For example, for classification problems, you can use Cross-Entropy Loss, while for regression problems, you can use Mean Squared Error (MSE) Loss. The right choice will help your model learn better.
3. Metrics: Metrics are measurements used to evaluate model performance after training. These metrics provide additional information apart from the loss function and can be used to understand the extent to which your model has successfully solved the given problem. Some common metrics include accuracy for classification problems, Mean Absolute Error (MAE) or Root Mean Squared Error (RMSE) for regression problems, and various other metrics such as F1-score, Precision, Recall, etc., depending on the type of problem faced.

After the model has been successfully compiled, it is ready to be trained. Before entering the training stage, let's display a summary of the model we have created.

````python
model.summary()
````

**3. Training the Model**

In this stage, we will train the model using the training data created previously.
The `model.fit` function requires input data `X_train` and labels `y_train`, and trains the model for 100 epochs. 20% of the training data is used as validation data to monitor model performance and help prevent overfitting. The training results, including performance metrics such as accuracy and loss for each epoch, are stored in the `history` object, which can be used for further analysis and visualization of the model training process.

````python
# Train Model
history = model.fit(X_train, y_train, epochs=100, validation_split=0.2)
````

**4. Model Evaluation**

After the model is trained, we then evaluate the model to determine its performance.

````python
# Evaluasi Model
loss, accuracy = model.evaluate(X_test, y_test)
print(f'Test Loss: {loss:.4f}')
print(f'Test Accuracy: {accuracy:.4f}')
````

We can also display accuracy and loss graphs for training and validation data during the training process as follows.

````python
plt.figure(figsize=(10,4))

# Plot akurasi
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')

# Plot loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')

plt.tight_layout()
plt.show()
````

Also display its Classification Report as follows.

````python
from sklearn.metrics import classification_report

# Prediksi pada data test
y_pred = model.predict(X_test)
y_pred_classes = (y_pred > 0.5).astype(int)
print(classification_report(y_test, y_pred_classes))
````

Test the model on new data

````python
# Data baru untuk diuji (contoh 2 titik data)
X_new = np.array([[0.5, 0.3], [1.2, -0.8]])
proba = model.predict(scaler.transform(X_new))
kelas = (proba > 0.5).astype(int)

for d, p, k in zip(X_new, proba, kelas):
    print(f"Data: {d}, Prob kelas 1: {p[0]:.4f}, Prediksi: {k[0]}")
````

Congratulations! With this, we have successfully created a simple model for binary classification.
