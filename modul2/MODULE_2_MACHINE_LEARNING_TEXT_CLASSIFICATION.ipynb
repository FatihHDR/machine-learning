# MAIN TOPICS

In today’s digital age, the volume of text-based data is steadily increasing alongside the development of communication technology and social media. Text-based information is available in various forms, such as emails, social media comments, news articles, product reviews, and more. To effectively utilize this text data, techniques are needed that can automatically group or classify text. One commonly used solution is text classification.


## DEFINITION OF TEXT DATA

Text data is a collection of information presented in the form of text, such as sequences of words, sentences, or paragraphs that convey a specific meaning. This information can come from various sources, such as documents, articles, social media posts, and so on. Here are some examples:

Text data can take various forms and formats, depending on the context and application. Examples include :
1. Text-based documents, such as reports, books, or papers.
2. Digital communications, such as emails, instant messages, and tweets.
3. Transcripts of conversations or interviews.


## APPLICATIONS OF TEXT DATA PROCESSING

Text data is used in various sectors, such as finance, insurance, healthcare, and others. Here are some examples of text data processing applications :

**1. Spam Filters**

One of the most annoying things about email is spam. These spam filters scan the text in all the emails we receive and try to figure out what they mean to determine whether they’re spam or not.

**2. Sentiment Analysis**

Sentiment analysis can be used to gauge customer opinions, satisfaction levels, and market response to products by analyzing social media posts, customer reviews, and survey responses. This can help companies make decisions, particularly when formulating marketing strategies.

**3. Chatbot and Virtual Customer Support**

Text data processing enables chatbots to understand and respond to customer questions and comments in a conversational manner. This technology is widely used in customer service.


**4. Automatic Translation**

Machine translation is the process of automatically converting text from one language to another using a computer. This technique is very useful for overcoming language barriers in global communication.

**5. Automatic Summary**

Automatic summarization is the process of automatically generating a summary from a long text. The goal is to present the main information from the original document in a more concise form, without omitting key points.


## PREPROCESSING STAGES FOR TEXT DATA

Text data preprocessing is one of the key steps in preparing data for machine learning or natural language processing (NLP) models. Here are 10 common steps in the text data preprocessing process.


**1. Removing Punctuation and Special Characters**

Punctuation marks such as periods, commas, exclamation points, and special characters such as @, #, or numbers are often unnecessary for text processing. This is because these characters can introduce noise and interfere with the model’s learning process.

**2. Removing Extra Spaces**

These extra spaces usually appear due to typos or when copying text from another source.


**3. Converting Text to Lowercase**

Computers distinguish between uppercase and lowercase letters, so without converting to lowercase, “Data” ≠ “data” and “AI” ≠ “ai.”

**4. Tokenization**

Breaking text down into smaller units, usually words or sentences. Each word becomes a token that can be processed further.


**5. Stopwords Removal**
Stopwords are common words that frequently appear in language but have no significant meaning in the context of analysis, such as “which,” “and,” and “at.” These words are often removed to reduce noise.


**6. Stemming dan Lemmatization**

**a. Stemming**

The process of converting a word to its base form by removing prefixes or suffixes according to specific rules, regardless of the word’s linguistic structure.

**b. Lemmatization**

The process of converting a word into its base form (lemma) while taking into account the word’s grammatical structure and meaning.


**7. Feature Engineering**

In Natural Language Processing (NLP), vectors play a crucial role in transformation. Processed text is converted into a numerical representation so it can be used by machine learning models. This process is typically carried out using a vectorizer.

A vector is essentially a numerical representation of a word, phrase, or even an entire document. These vectors are not merely numbers; they also capture the meaning (semantics) and grammatical properties (syntax) of the text, allowing machines to perform mathematical operations on them. With this representation, algorithms can perform tasks such as classification, clustering, and prediction. Some common techniques used include:


**a. One-Hot Encoding:** Represents words as binary vectors.

**b. Bag of Words (BoW):** Represents text by counting the number of times a word appears in a document.


**c. Text Vectorization:** Converts raw text into a numerical representation (vector).


**d. Term Frequency-Inverse Document Frequency (TF-IDF):** Measures the importance of a word in a document relative to the corpus.

**e. CountVectorizer:**  A practical implementation of Bag of Words in scikit-learn.


**f. Word Embeddings:** Using numerical vectors to represent words, such as Word2Vec or GloVe.


**8. Data Splitting**

Split the dataset into training, validation, and test sets, if necessary.


**9. Normalization or Standardization**

In the context of text data, normalization is the process of converting non-standard words, abbreviations, or non-standard spellings into their standard or commonly used forms.


**10. Storage or Serialization**

The processed dataset can be saved in an appropriate format, such as CSV or JSON, or directly converted into an input vector for use in a model.

# **SIMPLE SIMULATION**

To classify text, we can use a simple model with Sklearn. In this simulation, we will create and train a model using a simple neural network, such as Naive Bayes. Here are the steps:


**1. Setting Up the Environment and Importing the Required Libraries**

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report
```

An RNN (Recurrent Neural Network) is a type of artificial neural network specifically designed to process sequential data, such as text, speech, music, or time series.

Unlike standard feedforward neural networks, which process inputs only once, RNNs have “memory”:


- The output from the previous step can affect the calculations in the next step.
- In other words, an RNN can remember contextual information from the past.


**2. Dataset Preparation**

Here is an example of a simple dataset.

```python
texts = [
    "I love 123 this movie !!!",
    "This is https://fufufafa.com an amazing film ???",
    "I hated the movie @sahroni",
    "This was the 000 worst film ever !!??",
    "What a great movie :-)",
    "This film is terrible 1312 !!!",
    "Absolutely fantastic @windah",
    "I really 123 enjoyed this film https://soundhoreg.com",
    "I did not like this movie #badmovie",
    "It was a total waste of time !!! 000",
    "One of the best movies ever :-)",
    "This movie is boring ???",
    "Would recommend it to everyone https://apalah.com",
    "I will never watch this film again @greenpink"
]

labels = [
    "positive","positive","negative","negative","positive","negative","positive","positive",
    "negative","negative","positive","negative","positive","negative"
]
```

**3. Data Preprocessing**

```python
# Encode Label
le = LabelEncoder()
y = le.fit_transform(labels)

# Split data
X_train_txt, X_test_txt, y_train, y_test = train_test_split(texts, y, test_size=0.30, random_state=42, stratify=y)

# Step 1-6 bagian TAHAPAN PREPROCESSING PADA DATA TEXT
def custom_standardize(s):
    s = tf.strings.lower(s)
    s = tf.strings.regex_replace(s, r"http\S+|www\.\S+", " ")
    s = tf.strings.regex_replace(s, r"@\w+|#\w+", " ")
    s = tf.strings.regex_replace(s, r"[^a-z\s]", " ")
    s = tf.strings.regex_replace(s, r"\s+", " ")
    return tf.strings.strip(s)

# Vectorizer (ubah teks ke representasi angka)
VOCAB_SIZE = 2000
MAX_LEN    = 12

vectorizer = layers.TextVectorization(
    standardize=custom_standardize,
    max_tokens=VOCAB_SIZE,
    output_mode="int",
    output_sequence_length=MAX_LEN
)
vectorizer.adapt(np.array(X_train_txt))

# Ubah teks -> urutan indeks kata (siap ke Embedding)
X_train = vectorizer(np.array(X_train_txt)).numpy()
X_test  = vectorizer(np.array(X_test_txt)).numpy()
```

TextVectorization is a Keras layer for preprocessing text (cleaning, tokenization, integer encoding, padding) so that raw text can be used directly by a neural network model.

**4. Building & Training Models**

```python
# Model SimpleRNN
model = models.Sequential([
    layers.Embedding(VOCAB_SIZE, 32, input_length=MAX_LEN),
    layers.SimpleRNN(32),
    layers.Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()

# Training Model
history = model.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=25, batch_size=4, verbose=0
)
```

**5. Model Evaluation**

```python
# Evaluasi Model
y_proba = model.predict(X_test, verbose=0).ravel()
y_pred  = (y_proba >= 0.5).astype(int)
print("Akurasi :", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=le.classes_))
```

**6. Test Your Predictions on New Data**

```python
# Predikasi pada Data Baru
samples = [
    "I absolutely loved it",
    "This was awful and I regret watching it",
    "Not bad, could be better"
]
X_new = vectorizer(np.array(samples)).numpy()
preds = (model.predict(X_new, verbose=0).ravel() >= 0.5).astype(int)
for s, p in zip(samples, preds):
    print(f"- {s} -> {le.inverse_transform([p])[0]}")
```

**Congratulations!** We’ve successfully built a simple model for text classification. Keep in mind that the model’s accuracy may not yet be optimal because we’re using a small dataset; ideally, deep learning works more effectively with larger datasets.
