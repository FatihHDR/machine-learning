## CORE MATERIAL

### 1. Image
An image, or more commonly called a picture, is a visual representation of an object or scene that can be a drawing, photograph, or graphic. Examples include photos of cats, landscapes, or X-rays in hospitals. So, how do computers see images? Computers cannot see images like human eyes. So, computers store images as collections of pixels. The more pixels in an image, the clearer and sharper the image will be.

Images are divided into two types: color images and grayscale images. Color images have pixels that store red, green, and blue (RGB) color information. The combination of these three colors produces millions of other colors. Meanwhile, grayscale images, or commonly called grayscale, only have black, white, and shades of gray. Each pixel does not store color, but only the brightness level from dark to light. These images are often used in data analysis because they are simpler for computers to process.

### 2. Image Classification
Image classification is one of the main tasks in the field of computer vision. Its goal is to teach computers to recognize and group objects in an image. For example, when you see a photo of a dog, your brain automatically recognizes it as a dog. So, the image will be processed and labeled so that the computer can recognize and classify objects in an image.

image.png


Single-label classification is a method that classifies an object into only one category. For example, when you classify images of cats vs. dogs. The model must choose one of two labels, namely “cat” or “dog”. It is not possible for one photo to be labeled “cat” and “dog” at the same time. In essence, the result of single-label classification only has one correct output. Meanwhile, multi-label classification allows an object to have more than one category or label simultaneously.

### 3. CNN
Convolutional Neural Network (CNN) is a type of neural network specifically designed to process visual data, such as images. CNN has become one of the most successful methods in the field of computer vision due to its ability to recognize visual patterns automatically, without us having to specifically tell it what to look for. Simply put, CNN is inspired by the way the human brain works, so this architecture works like our eyes and brain when looking at an object. The human eye does not process the entire image at once, but rather recognizes an object by first identifying its small features—such as lines, corners, or colors, then combining these features into more complex shapes.

CNN has several main layers with specific functions that process data from the previous layer.

**1. Convolutional Layer**

The convolutional layer is the core layer of CNN that is useful for extracting features from images. It works by using a filter/kernel, which is a small matrix containing numbers. This filter/kernel is tasked with finding specific patterns in the image, such as edges, corners, or textures. The result of this layer is called a feature map.

In CNN, image data is generally represented in three dimensions: Height (H), Width (W), and Number of Channels (C). For example, an RGB image (3 color channels) measuring 5x5 pixels will have an input dimension of (5x5x3), where 3 is the number of color channels. To process this image, CNN uses a filter/kernel that also has three dimensions: height (K), width (K), and number of channels (C). It is also important to note that the number of channels must be the same as the number of channels in the input image. For example, for an RGB image (5 x 5 x 3), a commonly used kernel size is (3 x 3 x 3). In general, the process of the convolutional layer can be formulated as follows:

image.png

In this formula, Mi represents the size of the feature in the input, Wi is the weight matrix of the convolution kernel, f(. ) is a non-linear activation function (e.g., ReLU) and feature_surfaceout is the output of the feature. If we use the Keras library to create a convolutional layer, the code is as follows:



```
Input(shape=H, W, 3))
Conv2D(32, (3,3), activation='relu)
```

Input(shape=H,W,3)) is the definition of our input image. The number 3 here refers to the number of color channels (C) in the image, which is usually for RGB (Red, Green, Blue) images. This corresponds to the components Mi and C in the formula, where Mi is the feature data on each input channel (e.g., Red, Green, and Blue channels), and C is the total number of channels (in this case, 3), which becomes a limit for the sum operation (Σ) in the formula.

Meanwhile, in the second line, the Conv2D function is a function that performs two-dimensional convolution operations, where 32 is the number of filters or kernels that will be used, meaning that the convolution operation will produce 32 feature maps. Furthermore, the parameter (3,3) is the size of each kernel. So, in short, the code line is a practical way to implement the entire mathematical process described in the formula. The code Conv2D(32, (3, 3), activation='relu') will automatically perform the multiplication between the input (Mi) and the kernel (Wi), sum the results (B), and activate it with a function (f), then repeat it 32 times to produce 32 Feature Maps. **2. Pooling Layer**

The pooling layer will receive feature maps from the previous layer. The task of the pooling layer is to simplify the existing information (downsampling). This layer reduces the size of the feature map by taking the most important values from the feature map so that only the essence of the features is retained, thus the pooling layer will make the model more efficient in terms of computation and help it become less sensitive to small shifts in the position of objects in the image.

image.png

In general, the types of pooling layers that are often used are max pooling and average pooling. In short, max pooling is a type of pooling that takes the maximum value from each small window that slides across the feature maps. This will make the model less sensitive to small shifts (translations) in the position of objects. Meanwhile, average pooling is a type of pooling layer that calculates the average value of each small window that slides across the feature maps. Unlike max pooling which focuses on the most dominant features, average pooling provides a general overview or average of the features in each area. This method is suitable when evenly distributed details are more important than prominent single features. However, these two types of pooling are not necessarily the best; there are other better types of pooling, but these are the most commonly used types in CNN.

**3. Flatten Layer**

image.png

After passing through the convolutional layer and pooling layer, where more complex features are continuously built, the summarized information is then transformed into a single long data vector through the flatten layer. So, the flatten layer will prepare the data for the final stage in the form of a vector. This data vector, which now contains a numerical representation of all the important features that have been extracted, is then fed into the next layer.

For example, when we have a feature map from a pooling layer with a size of 3x3x5, the total elements or numerics in it are 3*3*5 = 45. When this data passes through the flatten layer, it will be converted into a one-dimensional vector with a length of 45. So, it can be said that the Flatten Layer is the data “preparation” stage. It takes all the feature information found by the CNN and presents it in a format that can be understood by the classification layer, allowing the model to make accurate final predictions.

4. FULLY CONNECTED LAYER

image.png

The fully connected layer, or commonly called the dense layer, is a layer that will analyze the relationships among all existing features to make a final decision. This layer only accepts one-dimensional input, so the flatten layer is very necessary in CNN models. Analogously, if the flatten layer is a bridge, then the fully connected layer is the brain that makes classification decisions. By combining all the features that have been extracted, the fully connected layer uses trained weights and biases to identify complex patterns and make classification decisions. The output of this layer, especially at the end of the CNN architecture, is often probabilities for each class, indicating how confident the model is about each possible category, such as "cat" or "dog".

For a clearer understanding, observe the CNN diagram below.

image.png

In the diagram above, the input image given is a “tweety” image. The image will then be processed through the convolutional layer to extract features present in the image. In this layer, a small matrix called a kernel/filter will slide across the entire image. Each time the kernel/filter slides, it will perform calculations that produce a single value. These values are collected into a feature map.

Next, the feature map will be taken to the pooling layer to summarize the contents of the data in the feature map. The result is usually called a pooled feature map. Its size is smaller, but the essence of the features is maintained.

In the diagram above, after a pooling layer is applied, a convolutional and another pooling layer are applied again. These two layers are usually repeated several times, the purpose of which is to reprocess the results of the previous layers, where simple features are combined into more complex features at each stage.

After the feature extraction process is complete, the results enter the classification stage. Here, the final feature maps are transformed into a single long data row through the Flatten Layer. This data is then fed into the Fully Connected Layer, which is responsible for combining all the extracted feature information. This layer will learn patterns from these features to make a final decision. This decision is represented in the Output Layer in the form of probabilities. Each neuron in this layer represents a category, and its probability value indicates how confident the model is that the image belongs to that category.

As an example from the diagram, after going through the entire process, the model predicts that the image is "Tweety" with a probability of 70%, which is the highest confidence. This step-by-step flow allows CNN to "learn" and "see" images in a way similar to humans, from small features to overall object recognition.

````python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
````

* ImageDataGenerator will function to perform image data augmentation and to load image data from a directory into the model in the correct format.
* Sequential is the simplest and most commonly used API model. So, sequential will be useful for building a neural network model by stacking several layers one by one linearly. Sequential will act as a container to build the model.

### Preprocessing data

Next, initialize **ImageDataGenerator** to perform image preprocessing and create data generators for training and validation.

````python
datagen = ImageDataGenerator(rescale=1./255)

train_generator = datagen.flow_from_directory(
    base_dir,
    target_size=(150, 150),
    batch_size=32,
    class_mode='binary')

validation_generator = datagen.flow_from_directory(
    base_dir,
    target_size=(150, 150),
    batch_size=32,
    class_mode='binary')
````

Explanation of arguments for train_generator and validation_generator:
* base_dir: This is the first argument given, which is a variable pointing to the main directory where the image dataset is located ('animals'). flow_from_directory will look for subdirectories within base_dir. In the case of this dataset, there are 'cat' and 'dog' subdirectories. This method will automatically consider the subdirectory names as class names.

* target_size=(150, 150): This argument specifies the size (height, width) to which all images will be resized. Although the original images may have different resolutions, the generator will resize them to 150x150 pixels before feeding them into the model. This is important because CNN models require image inputs with consistent sizes.

* batch_size=32: This sets the batch size of data that will be generated by the generator. During training, the model does not process all images at once, but rather processes them in small groups called batches. A batch size of 32 means that each time the model requests data from the generator, the generator will provide 32 images and their corresponding labels.

* class_mode='binary': This argument specifies the type of labels to be generated. Since we are doing binary classification (pizza vs not pizza, i.e., 2 classes), we use 'binary'. This will make the generator produce binary labels (0 or 1) for each image, corresponding to its class. If you have more than two classes, you would use 'categorical'

So, what's the difference between the two generators if they have the same arguments? Because we only have two dataset subfolders, namely 'cat' and 'dog', without separating them into train, test, and validation sets. Thus, the arguments used are the same. Now, if two generators have the same arguments, functionally there is no difference in the data source they access. Both will read images from the same directory (base_dir), with the same target size, batch size, and class mode. The main difference lies in their purpose during the model training process.

* train_generator is used in the model.fit() method to provide batches of data that the model will use to learn and update its weights.

* validation_generator is also used in the model.fit(), but to provide batches of data used to monitor model performance during training without updating the model's weights.

This validation result helps identify if the model is starting to overfit—for example, performing well on training data but declining on validation data. However, if both generators read from the same directory, it means the validation process is performed on the exact same data as the training data. This can give a misleading picture of validation performance and mask potential overfitting problems. In proper practice, validation_generator should read from a separate validation data set that has never been used by the model during training.

### Build CNN model

To build a CNN model, we need a conceptual understanding of how the model will process images as explained in point 3: CNN, about the layers needed to build a model that will identify visual patterns to distinguish between the two images. This process is largely done through a feature extraction stage consisting of a combination of convolution and pooling layers. Before building the model, let's imagine what layers our model needs to classify pizza and non-pizza images. First, we need a convolution layer to extract basic features such as lines, edges, or textures from the image. After that, we need a pooling layer to reduce the size of the representation while maintaining important information. This combination of convolution and pooling is usually repeated several times so that the model can recognize increasingly complex patterns, such as the characteristic shape of pizza or the characteristics of a non-pizza image.

Next, the results of the feature extraction are processed through a flatten layer to be converted into a one-dimensional vector, then passed to a dense layer so that the model can make decision-making processes. Finally, we need an output layer with sigmoid activation to produce the probability of whether the given image is pizza or not. If written in code, it will look like the following code:

````python
model = Sequential([
    Conv2D(16, (3, 3), activation='relu', input_shape=(150, 150, 3)),
    MaxPooling2D((2, 2)),
    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.summary()
````

You can modify the CNN layers according to your dataset needs.

* Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)): This is the first layer in the model, which is a 2D convolutional layer (Conv2D).

* 32: The number of convolutional filters (kernels) in this layer. Each filter will learn to detect different features in the image.

* (3, 3): The size of each convolutional filter (height x width). Here, the filter is 3x3 pixels.

* activation='relu': The activation function used after the convolution operation. ReLU (Rectified Linear Unit) is a commonly used and effective activation function.

* input_shape=(150, 150, 3): This only needs to be specified in the first layer of a Sequential model. It defines the expected input shape for the model: 150 pixels high, 150 pixels wide (adjust according to your image data size), and 3 color channels (for RGB images).

* MaxPooling2D((2, 2)): This is a 2D max pooling layer.
(2, 2): The size of the pooling window. This layer takes the maximum pixel value from each 2x2 pixel block in the previous layer's output. This helps reduce the spatial dimensions (height and width) of the output, reducing the number of parameters, and helping the model become more robust to small variations in the position of features in the image.

* Flatten(): This layer transforms the output of the previous layer (which is 3D: height, width, channels) into a 1D vector (one-dimensional). This is necessary because the subsequent Dense layer (fully connected layer) expects 1D input.

* Dense(64, activation='relu'): This is the first Dense (fully connected) layer.
64: The number of units (neurons) in this layer. Each neuron is connected to all neurons in the previous layer.

* Dense(1, activation='sigmoid'): This is the last Dense layer (output layer). Here, 1: The number of units in the output layer. Since this is a binary classification problem (pizza vs. not pizza), we only need one output unit.

* activation='sigmoid': The sigmoid function produces an output between 0 and 1, which can be interpreted as the probability of the positive class (e.g., the probability that the image is a dog).

* model.summary(): This method prints a summary of the model, including the name of each layer, its output shape, and the number of trainable parameters in each layer. This is very useful for understanding the model structure and the number of parameters that need to be learned. Thus, from the code above, we can see a model overview like this:

````python
model.summary()
````

So, there is actually no rigid rule that determines the exact number of layers that must be used. Instead, the number of layers in a CNN model highly depends on several factors:

* Problem complexity: Classifying pizza and non-pizza is a relatively simple case compared to classifying thousands of different objects. For more complex problems, usually more layers are needed to extract finer features and differentiate between classes.
* Image data size and quality: For example, if the image data used is small or has low resolution and a lot of noise, a model with too many layers will tend to overfit (the model memorizes the training data too much and its performance will be poor on new data). Conversely, if the data we have is large and high-quality, more layers can help the model learn richer representations.
* Availability of Computational Resources: Models with more layers require more computational power and training time. If resources are limited, a simpler architecture with fewer layers might be more practical.

### Compile model

So that the model can be trained, we must compile (prepare/set up the model before training) it:

````python
from tensorflow.keras.optimizers import Adam

model.compile(optimizer=Adam(),
              loss='binary_crossentropy',
              metrics=['accuracy'])
````

As explained in the previous module, we will use the Adam optimizer. Loss is a function that will be used by the model during training to measure how good or how bad the model's performance is on the training data. The goal is to minimize the value of the loss function. Meanwhile, 'binary_crossentropy' is the name of the specific loss function used here. Binary crossentropy is a very common and recommended loss function for binary classification problems (when you have two classes, such as pizza vs. non-pizza). This function measures the difference between the probability distribution predicted by the model (the output of the sigmoid layer, which is a probability between 0 and 1) and the actual probability distribution (binary labels 0 or 1). The smaller the binary_crossentropy value, the closer the model's predictions are to the actual labels. So, by setting loss='binary_crossentropy', we tell the model to use this function as a measure of "error" during training, and the optimization process (specified by optimizer=Adam()) will try to reduce this error.

### Train model

If the model is ready, we just need to train the model with the dataset we prepared earlier.

````python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

epochs = 7
history = model.fit(
    train_generator,
    epochs=epochs,
    validation_data=validation_generator,
    callbacks=[EarlyStopping(monitor='val_loss', patience=5), ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, verbose=1, min_lr=1e-7)]

)
````

model.fit() will run training using the data we provided in the specified parameters such as training data (train_generator), number of epochs, validation data (validation_generator), and callbacks. What are callbacks? Callbacks are processes in model training that allow certain functions or actions at specific points during the training process. Or more clearly, a callback is a function or object that we give to model.fit() so that this callback runs automatically at certain moments during model training.

For example, we have 15 epochs to train the model to achieve its highest accuracy, but it turns out that after the model is trained until the last epoch, the model's accuracy actually decreases. After performing visual analysis, it turns out that the model achieved its highest accuracy at the 7th epoch. But, it's not certain that if we reduce the epochs according to the peak accuracy, the model will have the same accuracy, and it would even be a waste of time if we re-train the model. This is where callbacks are very useful. Here are some examples of commonly used callbacks:

* EarlyStopping: Automatically stops training if the model's performance does not improve, so overfitting will not occur.

* ModelCheckpoint: Saves the best weights when the model's performance is at its peak, even if training continues until the end.

* ReduceLROnPlateau: Useful for lowering the learning rate if the model's performance cannot improve. It's like, when the model is stuck at that performance, this function will give a smaller learning rate so that it can find the optimal point again.

There are many other types of callbacks. For a more complete explanation and their parameters, you can read the following documentation: [KERAS CALLBACK API](https://keras.io/api/callbacks/)

### Evaluate model

### Image Classification Function

If the model has been evaluated using metrics, we will perform classification using the model we trained. Use the following code:

````python
import numpy as np
from tensorflow.keras.preprocessing import image
import matplotlib.pyplot as plt

'''
Tentukan parameter sesuai kebutuhan kalian,
target size disesuaikan dengan ukuran citra saat kalian melatih model
'''
def classify_image(image_path, model, target_size=(150, 150)):
    img = image.load_img(image_path, target_size=target_size) #Memuat gambar dan mengubah ukuran sesuai dengan target size
    img_array = image.img_to_array(img) #Mengubah objek gambar menjadi array numpy agar sesuai dengan input yang dibutuhkan CNN
    img_array = np.expand_dims(img_array, axis=0) #Menambahkan dimensi baru di awal array karena model membutuhkan input dalam bentuk batch
    img_array /= 255.0 #Normalisasi nilai piksel pada gambar (sesuai dengan preprocessing yang dilakukan saat melatih model)

    prediction = model.predict(img_array)
    probability = prediction[0][0] #Membuat prediksi pada array citra yang sudah diproses

    #Mengambil nilai probabilitas dari hasil prediksi
    if probability > 0.5:
        predicted_class = 'pizza'
        confidence = probability * 100
    else:
        predicted_class = 'not pizza'
        confidence = (1 - probability) * 100


    plt.imshow(img)
    plt.title(f"Predicted: {predicted_class} ({confidence:.2f}%)")
    plt.axis('off')
    plt.show()

    return predicted_class
````
